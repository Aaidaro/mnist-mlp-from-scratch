```python
import gzip, pickle, os, numpy as np, matplotlib.pyplot as plt, pandas as pd
path='../data/mnist.pkl.gz'
with gzip.open(path,'rb') as f:
    data=pickle.load(f, encoding='latin1')
type(data), len(data)
```

```python
for i,part in enumerate(data):
    print('index=', i, 'type=', type(part),'len=', len(part))

    if isinstance(part, tuple):
        for j,arr in enumerate(part):
            print(' ',
                  j,
                  type(arr),
                  getattr(arr,'shape',None),
                  getattr(arr,'dtype',None)
                  )
    else:
        print(part[:2])
```

```python
(train_X, train_y), (valid_X, valid_y), (test_X, test_y) = data

print('train', train_X.min(),train_X.max())
print('Val', valid_X.min(), valid_X.max())
print('Test', test_X.min(), test_X.max())

unique_train, counts_train = np.unique(train_y, return_counts=True)
print(dict(zip(unique_train, counts_train)))

unique_val, counts_val = np.unique(valid_y, return_counts=True)
print(dict(zip(unique_val, counts_val)))

unique_test, counts_test =np.unique(test_y, return_counts=True)
print(dict(zip(unique_test,counts_test)))
```
